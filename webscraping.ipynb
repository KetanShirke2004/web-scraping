```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'} 
url='https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav'
webpage=requests.get(url,headers=headers)
```

```python
webpage
```

```python
soup=BeautifulSoup(webpage.content,'html.parser')
```

```python
company=soup.find_all('div',class_="companyCardWrapper")
```

```python
len(company)
```

```python
name=[]
rating=[]
about=[]
salaries=[]


for i in company:
    name.append(i.find('h2').text.strip())
    rating.append(i.find('div',class_="rating_text").text.strip())
    about.append(i.find('span',class_="companyCardWrapper__interLinking").text.strip())
    salaries.append(i.find('span',class_="companyCardWrapper__ActionCount").text.strip())

df=pd.DataFrame({"Name":name,"Rating":rating,"About":about,"Salaries":salaries})
df
    
```

```python
df.shape
```

```python
final_dfs = []
for page in range(1, 8):
    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
    url = f'https://www.ambitionbox.com/list-of-companies?page={page}&campaign=desktop_nav'
    webpage = requests.get(url, headers=headers)
    soup = BeautifulSoup(webpage.content, 'html.parser')
    company = soup.find_all('div', class_="companyCardWrapper")
    
    name = []
    rating = []
    about = []
    salaries = []


    for comp in company:
        name.append(comp.find('h2').get_text(strip=True) if comp.find('h2') else "")
        rating.append(comp.find('div', class_="rating_text").get_text(strip=True) if comp.find('div', class_="rating_text") else "")
        about.append(comp.find('span', class_="companyCardWrapper__interLinking").get_text(strip=True) if comp.find('span', class_="companyCardWrapper__interLinking") else "")
        salaries.append(comp.find('span', class_="companyCardWrapper__ActionCount").get_text(strip=True) if comp.find('span', class_="companyCardWrapper__ActionCount") else "")

    df = pd.DataFrame({"Name": name, "Rating": rating, "About": about, "Salaries": salaries})
    final_dfs.append(df)

final = pd.concat(final_dfs, ignore_index=True)
```
